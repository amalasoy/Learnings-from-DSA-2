```python
def iqr_cap(series, factor=1.5):
  q1 = series.quantile(0.25)
  q3 = series.quantile(0.75)
  iqr = q3-q1
  lower = q1-factor*iqr
  upper = q3+factor*iqr

  capped= series.clip(lower,upper)
  return capped
```

```python
df['Age of House (in Years)_capped'] = iqr_cap(df['Age of House (in Years)'])
```

```python
df['Sale Price_capped'] = iqr_cap(df['Sale Price'])
df['Basement Area (in Sqft)_capped'] = iqr_cap(df['Basement Area (in Sqft)'])
df['Renovated Year_capped'] = iqr_cap(df['Renovated Year'])
df['Flat Area (in Sqft)_capped'] = iqr_cap(df['Flat Area (in Sqft)'])
```
